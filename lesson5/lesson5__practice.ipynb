time-spend-alone is fully float type so we can use mean method
as well as these featrues: social event attandance, going outside, post-frequency and friends-circle-size

there is only 2 features that have str type values and it still have missing points so in these parts we can use ether mode method or fixed one if we dont wanna use drop to clear one by one

```
df['Time_spent_Alone'].fillna(df['Time_spent_Alone'].mean())
```

```
df['Social_event_attendance'].fillna(df['Social_event_attendance'].median())
```

```
df['Going_outside'].fillna(df['Going_outside'].mean())
```

```
df['Post_frequency'].fillna(df['Post_frequency'].mode()[0])
```

```
df['Friends_circle_size'].fillna(df['Friends_circle_size'].mean())
```

```
df['Stage_fear'].fillna(df['Stage_fear'].mode()[0])
```

```
df['Drained_after_socializing'].fillna('Yes')
```

nice all done with empty values, now we can deal with encoding
